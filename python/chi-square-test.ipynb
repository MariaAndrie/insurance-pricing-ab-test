```python
import scipy
print(scipy.__version__)
```

```python
from scipy.stats import chi2_contingency
```

```python
# Contingency table:
# [converted, not converted]
# A_basic vs B_premium pricing groups

contingency = [
    [4673, 20177],   # A_basic
    [3817, 21034]    # B_premium
]
```

```python
print(contingency)
```

```python
# Run chi-square test to evaluate whether
# the difference in conversion rates is statistically significant

chi2, p, dof, expected = chi2_contingency(contingency)
```

```python
# Print test results

print(f"chi-square: {chi2:.2f}")
print(f"p-value: {p:.10f}")
```

A chi-square test confirmed that the difference in conversion rates between pricing groups was statistically significant (χ² = 103.88, p < 0.001), suggesting that pricing had a measurable impact on customer behavior.
